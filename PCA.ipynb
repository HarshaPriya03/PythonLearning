```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler

# Load dataset
digits = load_digits()
X = digits.data

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Covariance matrix
cov_mat = np.cov(X_scaled.T)

# Minimum covariance value
print("Minimum covariance:", cov_mat.min())
```

```python
from sklearn.decomposition import PCA

# PCA
pca = PCA()
pca.fit(X_scaled)      # You can also use X, but PCA is generally applied on scaled data.

# Explained variance ratio
print(pca.explained_variance_ratio_)

# Maximum explained variance ratio
print("Maximum explained_variance_ratio:",
      pca.explained_variance_ratio_.max())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.decomposition import PCA
from sklearn.metrics import root_mean_squared_error

# Read dataset
data = pd.read_csv("D:\LearningConcepts\intelipaat\ML models assignments\PCA Quiz\GYM.csv")

X = data.drop("number_people", axis=1)
y = data["number_people"]

# Model 1
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.2,random_state=24)

model1=LinearRegression()
model1.fit(X_train,y_train)

pred1=model1.predict(X_test)

rmse1=root_mean_squared_error(y_test,pred1)

print("RMSE Model1:",rmse1)

# PCA
pca=PCA(n_components=4)

X_pca=pca.fit_transform(X)

X_train,X_test,y_train,y_test=train_test_split(
    X_pca,y,test_size=0.2,random_state=24)

model2=LinearRegression()
model2.fit(X_train,y_train)

pred2=model2.predict(X_test)

rmse2=root_mean_squared_error(y_test,pred2)

print("RMSE Model2:",rmse2)

if rmse2<rmse1:
    print("Model 2 performance is better than Model 1")
elif rmse2==rmse1:
    print("No change in performance")
else:
    print("Model 2 performance is not better than Model 1")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset
data = pd.read_csv("D:\LearningConcepts\intelipaat\ML models assignments\PCA Quiz\GYM.csv")

# Independent features
X = data.iloc[:, :-1]

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Variance before PCA
var_before = np.var(X_scaled, axis=0, ddof=1).sum()

# PCA with 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Reconstruct the data
X_reconstructed = pca.inverse_transform(X_pca)

# Variance after reconstruction
var_after = np.var(X_reconstructed, axis=0, ddof=1).sum()

# Difference
difference = var_before - var_after

print("Variance Before:", round(var_before, 2))
print("Variance After:", round(var_after, 2))
print("Difference:", round(difference, 2))
```
