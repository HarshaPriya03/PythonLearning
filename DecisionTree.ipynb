```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Read data
df = pd.read_csv(r"D:\LearningConcepts\intelipaat\ML models assignments\Decision Tree Quiz\Pokemon.csv")

# Fill missing values
df["Type 2"] = df["Type 2"].fillna(df["Type 2"].mode()[0])

# Label Encode categorical columns
le = LabelEncoder()

for col in ["Name", "Type 1", "Type 2"]:
    df[col] = le.fit_transform(df[col])

# Target
y = df["Legendary"]

# Features
X = df.drop("Legendary", axis=1)

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=1
)

# Model
dt = DecisionTreeClassifier(random_state=1)
dt.fit(X_train, y_train)

# Prediction
y_pred = dt.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from scipy.stats import entropy

data = pd.read_csv(r"D:\LearningConcepts\intelipaat\ML models assignments\Decision Tree Quiz\diabetes.csv")

# ---------------- Q15 ----------------
bp = data["BloodPressure"].copy()

bp_cat = pd.cut(
    bp,
    bins=[-np.inf,79,120,np.inf],
    labels=["Low","Normal","High"]
)

p = bp_cat.value_counts(normalize=True)
print("Q15 Entropy =", entropy(p, base=2))

# ---------------- Q17 ----------------
def calc_entropy(col):
    p = data[col].value_counts(normalize=True)
    return entropy(p, base=2)

print("\nQ17")
for c in data.columns[:-1]:
    print(c, calc_entropy(c))

# ---------------- Q18 ----------------
X = data.drop("Outcome", axis=1)
y = data["Outcome"]

tree = DecisionTreeClassifier(
    criterion="entropy",
    random_state=0,
    max_depth=1
)

tree.fit(X,y)

print("\nQ18 Root Feature =", X.columns[tree.tree_.feature[0]])

# ---------------- Q19 ----------------
print("\nQ19 Gini")
for c in data.columns[:-1]:
    p = data[c].value_counts(normalize=True)
    gini = 1 - np.sum(p**2)
    print(c, gini)

# ---------------- Q21 ----------------
df = data[data.Glucose != 0]
df = df[df.Glucose <= 75]

X = df[["Glucose"]]
y = df["Outcome"]

tree = DecisionTreeClassifier(random_state=0)
tree.fit(X,y)

print("\nQ21 Leaf =", tree.tree_.feature[0] == -2)

# ---------------- Q22 ----------------
X = data[["Pregnancies","BloodPressure"]]
y = data["Outcome"]

X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2,random_state=24
)

dt = DecisionTreeClassifier(random_state=8)
dt.fit(X_train,y_train)

pred = dt.predict(X_test)

cm = confusion_matrix(y_test,pred)

TN,FP,FN,TP = cm.ravel()

precision = TP/(TP+FP)

print("\nQ22")
print(cm)
print("Precision =", precision)

# ---------------- Q23 ----------------
X = data[["Insulin","BMI"]]

X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2,random_state=24
)

dt = DecisionTreeClassifier(random_state=30)
dt.fit(X_train,y_train)

pred = dt.predict(X_test)

cm = confusion_matrix(y_test,pred)

TN,FP,FN,TP = cm.ravel()

recall = TP/(TP+FN)

print("\nQ23")
print(cm)
print("Recall =", recall)
```
