Use .loc when you know the specific text labels or dates of your data. Use .iloc when you want to quickly grab data based on its physical location in the table (e.g., "get the first 5 rows and the last 2 columns")

```python
import pandas as pd

data = {'Age': [25, 30, 35], 'City': ['New York', 'Paris', 'Tokyo']}
df = pd.DataFrame(data, index=['Alice', 'Bob', 'Charlie'])
df
```

```python
df.loc['Alice']  # Returns the row for Alice
```

```python
df.iloc[0]       # Returns the 1st row (Alice), since it is at index 0
```

```python
df.loc['Alice':'Bob']  
# Returns BOTH Alice and Bob
```

```python
df.iloc[0:2]           
# Returns positions 0 and 1 (Alice and Bob). Position 2 is excluded
```

```python
df.loc['Bob', 'City']  # Returns 'Paris'
```

```python
df.iloc[1, 1]          # Returns 'Paris' (2nd row, 2nd column)
```

```python
df.loc[df['Age'] > 28]  # Returns rows for Bob and Charlie
```
